```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from IPython.display import display

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
PROJECT_DIR = Path.cwd()

# If notebook is running inside notebooks folder
if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

DATA_FILE = PROJECT_DIR / "data" / "sensor.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project directory:", PROJECT_DIR)
print("Dataset path:", DATA_FILE)
print("Output directory:", OUTPUT_DIR)
```

```python
if not DATA_FILE.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_FILE}")

df = pd.read_csv(DATA_FILE)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Dataset information:")
df.info()

print("\nDataset columns:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nStatistical description:")
display(df.describe())
```

```python
print("Machine status distribution:")

display(
    df["machine_status"].value_counts().to_frame("Frequency")
)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="machine_status",
    order=df["machine_status"].value_counts().index
)

plt.title("Machine Condition Distribution")
plt.xlabel("Machine Status")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python
normal_df = df[
    df["machine_status"]
    .astype(str)
    .str.strip()
    .str.upper() == "NORMAL"
].copy()

print("Normal records:", normal_df.shape[0])

display(normal_df.head())
```

```python
sensor_columns = [
    column for column in normal_df.columns
    if column.lower().startswith("sensor_")
]

print("Total sensor columns:", len(sensor_columns))
print(sensor_columns)
```

```python
normal_df = normal_df.drop_duplicates()

for column in sensor_columns:
    normal_df[column] = pd.to_numeric(
        normal_df[column],
        errors="coerce"
    )

normal_df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

empty_columns = [
    column for column in sensor_columns
    if normal_df[column].isna().all()
]

normal_df = normal_df.drop(columns=empty_columns)

sensor_columns = [
    column for column in sensor_columns
    if column not in empty_columns
]

for column in sensor_columns:
    normal_df[column] = normal_df[column].fillna(
        normal_df[column].median()
    )

print("Cleaning completed!")
print("Removed empty columns:", empty_columns)
print("Remaining sensors:", len(sensor_columns))
print("Cleaned dataset shape:", normal_df.shape)

display(normal_df.head())
```

```python
normal_output = OUTPUT_DIR / "normal_records.csv"

normal_df.to_csv(normal_output, index=False)

print("Normal records saved successfully!")
print(normal_output)
```

```python
sensor_summary = normal_df[sensor_columns].describe().T

sensor_summary["Unique_Values"] = normal_df[
    sensor_columns
].nunique()

sensor_summary["Mode"] = normal_df[
    sensor_columns
].mode().iloc[0]

display(sensor_summary)

sensor_summary.to_csv(
    OUTPUT_DIR / "normal_sensor_frequency.csv"
)

print("Sensor statistics saved!")
```

```python
selected_sensors = sensor_columns[:12]

fig, axes = plt.subplots(
    4, 3,
    figsize=(18, 16)
)

axes = axes.flatten()

for i, column in enumerate(selected_sensors):

    sns.histplot(
        normal_df[column],
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(f"Distribution of {column}")
    axes[i].set_xlabel("Sensor Value")
    axes[i].set_ylabel("Frequency")

plt.suptitle(
    "Sensor Value Distributions - Normal Condition",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "normal_analysis_graphs.png",
    dpi=300
)

plt.show()
```

```python
correlation = normal_df[sensor_columns].corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    annot=False,
    center=0
)

plt.title("Sensor Correlation Heatmap - Normal Condition")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "normal_sensor_correlation.png",
    dpi=300
)

plt.show()
```
